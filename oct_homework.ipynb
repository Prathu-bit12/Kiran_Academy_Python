```python
# Program to mask sensitive digits of a string (keeping only last 4 visible)
old = "4548798451"
s = len(old)-4
new = ""
for ch in range(len(old)):
    if ch < s:
        new += "*"
    else:
        new += old[ch]
print("old number:", old)
print("new number:", new)
```

```python
# Reverse a string
str1 = "Python is a programming language"
new = ""
for i in range(len(str1)-1, -1, -1):
    new += str1[i]
print("old string:", str1)
print("new string:", new)
```

```python
# Palindrome check
str1 = "malayalam"
for i in range(len(str1)):
    if str1[i] != str1[len(str1)-1-i]:
        print("Not a palindrome")
        break
else:
    print("It is a palindrome")
```

```python
# Program to remove adjacent (consecutive duplicate) characters from a string
s = "zzbbcz"
new = ""
for i in s:
    if len(new)>0 and new[len(new)-1]==i:
        new = new[:-1]
    else:
        new += i

print("original string:", s)
print("new string:", new)
```

```python
# Program to check if two strings are anagrams of each other
s1 = "listen"
s2 = "silent"

if len(s1) != len(s2):
    print("It is not Anagram")
else:
    is_anagram = True
    for i in s1:
        
        count1 = 0
        count2 = 0
        
        for j in s1:
            if j == i:
                count1 += 1
                
        for k in s2:
            if k == i:
                count2 += 1
                
        if count1 != count2:
            is_anagram = False
            break

    if is_anagram:
        print("It is Anagram")
    else:
        print("It is not Anagram")
```

```python
# Program to check if a character is present in a string
s = "python"
target = "h"
found = False

for i in s:
    if i == target:
        found = True
        break

if found:
    print("Character mil gaya!")
else:
    print("Character nahi mila!")
```

```python
# Program to extract a substring from a string
s = "programming"
start = 3
end = 7
new_str = ""

for i in range(len(s)):
    if i >= start and i < end:
        new_str += s[i]

print("original string:", s)
print("substring:", new_str)  
```

```python
str1 = "Python 3 is cool"

vowels = 0
consonants = 0
digits = 0
spaces = 0

for i in str1:
    if i == " ":
        spaces += 1

    elif i >= "0" and i <= "9":
        digits += 1

    elif (
        i == "a"
        or i == "e"
        or i == "i"
        or i == "o"
        or i == "u"
        or i == "A"
        or i == "E"
        or i == "I"
        or i == "O"
        or i == "U"
    ):
        vowels += 1

    elif (i >= "a" and i <= "z") or (i >= "A" and i <= "Z"):
        consonants += 1

print("String:", str1)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
```
