```python
data = [5,4,5,4,6,7,8,4,3,5,3,4]
unique = []

for num in data:
    if num not in unique:
        unique.append(num)

print("Unique List:",unique)
```

```python
freq_ele = {}
for num in data:
    freq_ele[num] = freq_ele.get(num,0)+1

print("Frequency elements:",freq_ele)
```
