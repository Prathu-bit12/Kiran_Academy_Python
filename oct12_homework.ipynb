```python
#Find topper in physics,math and chemistry subjects from the given database of students marks.
db = {
    1: [78, 97, 86],
    2: [97, 94, 92],
    3: [78, 90, 98]
}

phy_topper = list(db.keys())[0]
math_topper = list(db.keys())[0]
chem_topper = list(db.keys())[0]


max_phy = db[phy_topper][0]
max_math = db[math_topper][1]
max_chem = db[chem_topper][2]

for k, v in db.items():
    
    if v[0] > max_phy:
        max_phy = v[0]
        phy_topper = k
        
   
    if v[1] > max_math:
        max_math = v[1]
        math_topper = k

    if v[2] > max_chem:
        max_chem = v[2]
        chem_topper = k


print(f"Physics Topper -> Roll Number {phy_topper} with marks {max_phy}")
print(f"Maths Topper -> Roll Number {math_topper} with marks {max_math}")
print(f"Chemistry Topper -> Roll Number {chem_topper} with marks {max_chem}")
```
